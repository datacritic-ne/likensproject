```python
def printing_results(df, rank):

    print(f'The average GF% of the {rank} team is: ', df['GF_percent'].mean())
    print(f'The {rank} team\'s average goal differential is: ', df['GD'].mean())
    print(f'The {rank} team averages the following standings points: ', df['PTS'].mean())
    print(f'The {rank} team\'s probability of winning the President\'s Trophy: ', df['President_Trophy'].mean())
    print(f'The {rank} team\'s probability of having the best record in their conferencce: ', df['Conf_Best_Record'].mean())
    print(f'The {rank} team\'s probability of winning their division: ', df['Division_Winner'].mean())
    print(f'The {rank} team\'s probability of making the playoffs: ', df['Made_Playoffs'].mean())
    print(f'The {rank} team\'s probability of winning 1 playoff round: ', df['Won_1_Round'].mean())
    print(f'The {rank} team\'s probability of winning 2 playoff rounds: ', df['Won_2_Rounds'].mean())
    print(f'The {rank} team\'s probability of winning 3 playoff rounds: ', df['Won_3_Rounds'].mean())
    print(f'The {rank} team\'s probability of winning the Stanley Cup: ', df['Won_Stanley_Cup'].mean())
    
```
