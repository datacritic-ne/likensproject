```python
import pandas as pd
```

```python
# This is code to calculate home-ice advantage and non-EN goals per game

# I want to do this over a few different intervals
# 1. Across all 20 seasons of the cap era
# 2. Across all seasons, excluding the 2012-13, 2019-20, and 2020-21 partial seasons
# 3. In 3-year moving averages, to capture changes in scoring environments and parity environments. So:
#    a. 2005-06, 2006-07, 2007-08
#    b. 2006-07, 2007-08, 2008-09
#    c. 2007-08, 2008-09, 2009-10
#    d. 2008-09, 2009-10, 2010-11
#    e. 2009-10, 2010-11, 2011-12
#    f. 2010-11, 2011-12, 2012-13
#    g. 2011-12, 2012-13, 2013-14
#    h. 2012-13, 2013-14, 2014-15
#    i. 2013-14, 2014-15, 2015-16
#    j. 2014-15, 2015-16, 2016-17
#    k. 2015-16, 2016-17, 2017-18
#    l. 2016-17, 2017-18, 2018-19
#    m. 2017-18, 2018-19, 2019-20
#    n. 2018-19, 2019-20, 2020-21
#    o. 2019-20, 2020-21, 2021-22
#    p. 2020-21, 2021-22, 2022-23
#    q. 2021-22, 2022-23, 2023-24
#    r. 2022-23, 2023-24, 2024-25
# 4. In 3-year moving averages without the partial seasons
#    a. 2010-11, 2011-12, 2013-14
#    b. 2011-12, 2013-14, 2014-15
#    c. 2017-18, 2018-19, 2021-22
#    d. 2018-19, 2021-22, 2022-23
```

```python
# The data files - these won't run until I'm using this notebook on a machine that has them saved locally

df_2006 = pd.read_csv("scraping_homeice_20052006.csv")
df_2007 = pd.read_csv("scraping_homeice_20062007.csv")
df_2008 = pd.read_csv("scraping_homeice_20072008.csv")
df_2009 = pd.read_csv("scraping_homeice_20082009.csv")
df_2010 = pd.read_csv("scraping_homeice_20092010.csv")
df_2011 = pd.read_csv("scraping_homeice_20102011.csv")
df_2012 = pd.read_csv("scraping_homeice_20112012.csv")
df_2013 = pd.read_csv("scraping_homeice_20122013.csv")
df_2014 = pd.read_csv("scraping_homeice_20132014.csv")
df_2015 = pd.read_csv("scraping_homeice_20142015.csv")
df_2016 = pd.read_csv("scraping_homeice_20152016.csv")
df_2017 = pd.read_csv("scraping_homeice_20162017.csv")
df_2018 = pd.read_csv("scraping_homeice_20172018.csv")
df_2019 = pd.read_csv("scraping_homeice_20182019.csv")
df_2020 = pd.read_csv("scraping_homeice_20192020.csv")
df_2021 = pd.read_csv("scraping_homeice_20202021.csv")
df_2022 = pd.read_csv("scraping_homeice_20212022.csv")
df_2023 = pd.read_csv("scraping_homeice_20222023.csv")
df_2024 = pd.read_csv("scraping_homeice_20232024.csv")
df_2025 = pd.read_csv("scraping_homeice_20242025.csv")
```

```python
# Concatenating the data files to calculate the interval estimates mentioned above

# All seasons in the cap era

df_allcap = pd.concat([df_2006, df_2007, df_2008, df_2009, df_2010, df_2011, df_2012, df_2013, df_2014, df_2015,
                      df_2016, df_2017, df_2018, df_2019, df_2020, df_2021, df_2022, df_2023, df_2024, df_2025
                      ], ignore_index=True)

# All cap-era full 82-game seasons

df_allcap_fullseasons = pd.concat([df_2006, df_2007, df_2008, df_2009, df_2010, df_2011, df_2012, df_2014, df_2015,
                      df_2016, df_2017, df_2018, df_2019, df_2022, df_2023, df_2024, df_2025], ignore_index=True)

# 2005-06 through 2007-08

df_20062008 = pd.concat([df_2006, df_2007, df_2008], ignore_index=True)

# 2006-07 through 2008-09

df_20072009 = pd.concat([df_2007, df_2008, df_2009], ignore_index=True)

# 2007-08 through 2009-10

df_20082010 = pd.concat([df_2008, df_2009, df_2010], ignore_index=True)

# 2008-09 through 2010-11

df_20092011 = pd.concat([df_2009, df_2010, df_2011], ignore_index=True)

# 2009-10 through 2011-12

df_20102012 = pd.concat([df_2010, df_2011, df_2012], ignore_index=True)

# 2010-11 through 2012-13

df_20112013 = pd.concat([df_2011, df_2012, df_2013], ignore_index=True)

# 2011-12 through 2013-14

df_20122014 = pd.concat([df_2012, df_2013, df_2014], ignore_index=True)

# 2012-13 through 2014-15

df_20132015 = pd.concat([df_2013, df_2014, df_2015], ignore_index=True)

# 2013-14 through 2015-16

df_20142016 = pd.concat([df_2014, df_2015, df_2016], ignore_index=True)

# 2014-15 through 2016-17

df_20152017 = pd.concat([df_2015, df_2016, df_2017], ignore_index=True)

# 2015-16 through 2017-18

df_20162018 = pd.concat([df_2016, df_2017, df_2018], ignore_index=True)

# 2016-17 through 2018-19

df_20172019 = pd.concat([df_2017, df_2018, df_2019], ignore_index=True)

# 2017-18 through 2019-20

df_20182020 = pd.concat([df_2018, df_2019, df_2020], ignore_index=True)

# 2018-19 through 2020-21

df_20192021 = pd.concat([df_2019, df_2020, df_2021], ignore_index=True)

# 2019-20 through 2021-22

df_20202022 = pd.concat([df_2020, df_2021, df_2022], ignore_index=True)

# 2020-21 through 2022-23

df_20212023 = pd.concat([df_2021, df_2022, df_2023], ignore_index=True)

# 2021-22 through 2023-24

df_20222024 = pd.concat([df_2022, df_2023, df_2024], ignore_index=True)

# 2022-23 through 2024-25

df_20232025 = pd.concat([df_2023, df_2024, df_2025], ignore_index=True)

# 2010-11 through 2013-14, without partial 2012-13 season

df_20112014_nopartial = pd.concat([df_2011, df_2012, df_2014], ignore_index=True)

# 2011-12 through 2014-15, without partial 2012-13 season

df_20122015_nopartial = pd.concat([df_2012, df_2014, df_2015], ignore_index=True)

# 2017-18 through 2021-22, without partial 2019-20 or 2020-21 seasons

df_20182022_nopartial = pd.concat([df_2018, df_2019, df_2022], ignore_index=True)

# 2018-19 through 2022-23, without partial 2019-20 or 2020-21 seasons

df_20192023_no partial = pd.concat([df_2019, df_2022, df_2023]), ignore_index=True)
```

```python
# Can I write a loop that calculates and prints sums and differences for each of the dataframes I've created?

list_of_dfs = [df_allcap, df_allcap_fullseasons, df_20062008, df_20072009, df_20082010, df_20092011, df_20102012, df_20112013, df_20112014_nopartial,
               df_20122014, df_20122015_nopartial, df_20132015, df_20142016, df_20152017, df_20162018, df_20172019, df_20182020, df_20182022_nopartial,
               df_20192021, df_20192023_nopartial, df_20202022, df_20212023, df_20222024, df_20232025]

for df in list_of_dfs:

    df['Home_Ice_Adv'] = df['HOME_GF'] - df['VISITOR_GF']
    df['Goals_Game'] = df['HOME_GF'] + df['VISITOR_GF']

    mean_homeice = df['Home_Ice_Adv'].mean()
    mean_goalsgame = df['Goals_Game'].mean()

    print(f'Home ice advantage in the file {df} is {mean_homeice}.')
    print(f'Average goals per game in the file {df} is {mean_goalsgame}')
```
